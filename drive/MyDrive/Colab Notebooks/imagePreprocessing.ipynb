```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

# Đọc ảnh bằng OpenCV (BGR)
img_bgr = cv2.imread("/content/drive/MyDrive/scoliosis.v1i.voc/test/IMG_1296_JPG.rf.c0e996c60beb1aebca32e3cc19db3093.jpg")

# Chuyển sang RGB để hiển thị đúng màu
img = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)


# 1. Chuyển ảnh sang grayscale
gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)  # dùng BGR để ra gray chính xác

# 2. Resize ảnh về (224,224)
resized = cv2.resize(img, (224, 224))

# 3. Chuẩn hóa (Normalization)
# chuyển sang float32 và scale về [0,1]
img = resized.astype(np.float32) / 255.0

# mean, std cho ImageNet (theo RGB)
mean = np.array([0.485, 0.456, 0.406])
std  = np.array([0.229, 0.224, 0.225])

# Chuẩn hóa từng kênh
img_normalized = (img - mean) / std

# 4. Lưu ảnh grayscale
cv2.imwrite("gray.jpg", gray)

print("Ảnh gốc:", img_bgr.shape)
print("Ảnh grayscale:", gray.shape)
print("Ảnh resize:", resized.shape)
print("Ảnh chuẩn hóa:", img_normalized.shape)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].imshow(img)
axes[0].axis("off")

axes[1].imshow(gray, cmap="gray")
axes[1].axis("off")

axes[2].imshow(img_normalized)
axes[2].axis("off")

plt.show()

```
