```python
import torch
import torchvision
from torchvision import transforms
from torch.utils.data import DataLoader

# Transform cho ảnh
transform = transforms.Compose([
    transforms.Resize((224, 224)),   # resize tất cả ảnh về 224x224
    transforms.ToTensor()            # chuyển thành tensor
])

# Dùng SpineDataset bạn đã định nghĩa
dataset = SpineDataset("/content/drive/MyDrive/scoliosis.v1i.voc/train", transform=transform)

# DataLoader
data_loader = DataLoader(dataset, batch_size=2, shuffle=True, collate_fn=lambda x: tuple(zip(*x)))

mean = 0.
std = 0.
total_images = 0

for images, _ in data_loader:  # dùng đúng tên data_loader
    # images lúc này là tuple => stack lại thành tensor
    images = torch.stack(images, 0)

    batch_samples = images.size(0)  # số ảnh trong batch
    images = images.view(batch_samples, images.size(1), -1)  # (batch, C, H*W)
    mean += images.mean(2).sum(0)   # mean theo (H*W)
    std += images.std(2).sum(0)     # std theo (H*W)
    total_images += batch_samples

mean /= total_images
std /= total_images

print("Mean:", mean)
print("Std:", std)
```
